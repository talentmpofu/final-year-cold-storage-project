````python
# ========================================================================
# YOLO11 MEDIUM - Fruit Classification Training
# Fast installation (~60 seconds) | Better accuracy than small/nano
# ========================================================================

# 1️⃣ Quick Installation
print("=" * 70)
print("YOLO11 SMALL TRAINING SETUP - 90 EPOCHS")
print("=" * 70)
print("\nInstalling packages (this will be quick!)...\n")

!pip install -q ultralytics roboflow

print("✓ Installation complete!")
print("=" * 70 + "\n")

# 2️⃣ Import libraries
import os
from roboflow import Roboflow
from ultralytics import YOLO
import yaml
from IPython.display import Image, display
import torch

# Check GPU availability
device = 'cuda' if torch.cuda.is_available() else 'cpu'
print(f"🖥️  Device: {device.upper()}")
if device == 'cuda':
    print(f"   GPU: {torch.cuda.get_device_name(0)}")
    print(f"   Memory: {torch.cuda.get_device_properties(0).total_memory / 1e9:.2f} GB")
print()

# 3️⃣ Download Dataset
print("=" * 70)
print("DOWNLOADING DATASET FROM ROBOFLOW")
print("=" * 70 + "\n")

rf = Roboflow(api_key="Ftw9gnZwYq00E7hBfVyo")
project = rf.workspace("talentmpofu").project("potatoes-uyno2-lpf5h")
version = project.version(1)
dataset = version.download("yolov8")  # Compatible with YOLO11

base_path = dataset.location
data_yaml_path = os.path.join(base_path, "data.yaml")

print(f"✓ Dataset downloaded to: {base_path}")
print(f"✓ Configuration file: {data_yaml_path}\n")

# 4️⃣ Load Dataset Configuration
print("=" * 70)
print("DATASET INFORMATION")
print("=" * 70 + "\n")

with open(data_yaml_path, 'r') as f:
    data_config = yaml.safe_load(f)

# Extract class information
if isinstance(data_config['names'], dict):
    class_names = [data_config['names'][i] for i in sorted(data_config['names'].keys())]
elif isinstance(data_config['names'], list):
    class_names = data_config['names']
else:
    class_names = list(data_config['names'].values())

num_classes = len(class_names)

print(f"📊 Number of classes: {num_classes}")
print(f"🏷️  Class names: {class_names}")

# Count images in each split
for split in ['train', 'valid', 'test']:
    img_dir = os.path.join(base_path, split, 'images')
    if os.path.exists(img_dir):
        img_count = len([f for f in os.listdir(img_dir) if f.endswith(('.jpg', '.jpeg', '.png'))])
        print(f"   {split.capitalize()}: {img_count} images")

print()

# 5️⃣ Initialize YOLO11 Medium Model
print("=" * 70)
print("MODEL INITIALIZATION")
print("=" * 70 + "\n")

# Load YOLO11 Small pretrained model
model = YOLO('yolo11n.pt')

print("✓ Model: YOLO11 Small (yolo11s)")
print("  - Pre-trained on COCO dataset")
print(f"  - Will be fine-tuned for {num_classes} fruit classes")
print("  - Faster training than medium, still good accuracy")
print()

# 6️⃣ Training Configuration
print("=" * 70)
print("TRAINING CONFIGURATION")
print("=" * 70 + "\n")

# Training hyperparameters - OPTIMIZED FOR FREE COLAB
EPOCHS = 90           # Number of training epochs
BATCH_SIZE = 16       # Batch size (adjust based on GPU memory)
IMAGE_SIZE = 640      # Input image size
PATIENCE = 15         # Early stopping patience
LEARNING_RATE = 0.001 # Initial learning rate
SAVE_PERIOD = 5       # Save checkpoint every 5 epochs

print("⚠️  FREE COLAB OPTIMIZATION:")
print(f"   - Using YOLO11 Small for faster training")
print(f"   - Training for {EPOCHS} epochs")
print(f"   - Early stopping at {PATIENCE} epochs if no improvement")
print(f"   - Checkpoints saved every {SAVE_PERIOD} epochs")
print(f"   - Estimated training time: 30-50 minutes")
print()

print(f"⚙️  Training Parameters:")
print(f"   - Epochs: {EPOCHS}")
print(f"   - Batch size: {BATCH_SIZE}")
print(f"   - Image size: {IMAGE_SIZE}x{IMAGE_SIZE}")
print(f"   - Initial LR: {LEARNING_RATE}")
print(f"   - Patience: {PATIENCE} epochs (early stopping)")
print(f"   - Optimizer: Adam")
print(f"   - Device: {device.upper()}")
print()

# 7️⃣ Train the Model
print("=" * 70)
print("STARTING TRAINING")
print("=" * 70)
print("\n🚀 Training in progress...\n")

results = model.train(
    data=data_yaml_path,
    epochs=EPOCHS,
    imgsz=IMAGE_SIZE,
    batch=BATCH_SIZE,
    name='yolo11n_fruits_classification',
    patience=PATIENCE,
    save=True,
    save_period=SAVE_PERIOD,  # Save checkpoints regularly
    device=device,
    workers=4,
    optimizer='Adam',
    lr0=LEARNING_RATE,
    lrf=0.01,
    momentum=0.937,
    weight_decay=0.0005,
    warmup_epochs=3,
    warmup_momentum=0.8,
    warmup_bias_lr=0.1,
    box=7.5,
    cls=0.5,
    dfl=1.5,
    pose=12.0,
    kobj=1.0,
    label_smoothing=0.0,
    nbs=64,
    hsv_h=0.015,
    hsv_s=0.7,
    hsv_v=0.4,
    degrees=0.0,
    translate=0.1,
    scale=0.5,
    shear=0.0,
    perspective=0.0,
    flipud=0.0,
    fliplr=0.5,
    mosaic=1.0,
    mixup=0.0,
    copy_paste=0.0,
    plots=True,
    verbose=True,
    cache=False  # Don't cache to save memory on free tier
)

print("\n" + "=" * 70)
print("✓ TRAINING COMPLETED!")
print("=" * 70 + "\n")

# 8️⃣ Validate the Model
print("=" * 70)
print("MODEL VALIDATION")
print("=" * 70 + "\n")

metrics = model.val()

print("📈 Validation Metrics:")
print(f"   - mAP@0.5: {metrics.box.map50:.4f} (50% IoU threshold)")
print(f"   - mAP@0.5:0.95: {metrics.box.map:.4f} (Average across IoU thresholds)")
print(f"   - Precision: {metrics.box.mp:.4f}")
print(f"   - Recall: {metrics.box.mr:.4f}")
print()

# 9️⃣ Test Predictions
print("=" * 70)
print("TESTING PREDICTIONS")
print("=" * 70 + "\n")

test_imgs_dir = os.path.join(base_path, 'valid', 'images')
if os.path.exists(test_imgs_dir):
    test_images = [f for f in os.listdir(test_imgs_dir) if f.endswith(('.jpg', '.jpeg', '.png'))]

    if test_images:
        # Test on first 3 images
        num_test = min(3, len(test_images))
        print(f"Running predictions on {num_test} test images...\n")

        for i in range(num_test):
            test_image_path = os.path.join(test_imgs_dir, test_images[i])
            print(f"[{i+1}/{num_test}] Predicting: {test_images[i]}")

            # Run inference
            results = model.predict(
                source=test_image_path,
                save=True,
                conf=0.25,
                iou=0.7,
                show_labels=True,
                show_conf=True,
                line_width=2
            )

        print(f"\n✓ Predictions saved to: runs/detect/predict/")
        print(f"   View the annotated images in that folder\n")

# 🔟 Export Model
print("=" * 70)
print("EXPORTING MODEL")
print("=" * 70 + "\n")

# Best model path
best_model_path = "runs/detect/yolo11s_solar_panel/weights/best.pt"
last_model_path = "runs/detect/yolo11s_solar_panel/weights/last.pt"

print("📦 Exporting to multiple formats...\n")

# Export to ONNX (most compatible for deployment)
try:
    onnx_path = model.export(format='onnx', simplify=True)
    print(f"✓ ONNX: {onnx_path}")
except Exception as e:
    print(f"⚠ ONNX export failed: {e}")

# Export to TorchScript
try:
    torchscript_path = model.export(format='torchscript')
    print(f"✓ TorchScript: {torchscript_path}")
except Exception as e:
    print(f"⚠ TorchScript export failed: {e}")

# Export to TensorFlow Lite (for mobile)
try:
    tflite_path = model.export(format='tflite')
    print(f"✓ TFLite: {tflite_path}")
except Exception as e:
    print(f"⚠ TFLite export failed: {e}")

print(f"\n✓ PyTorch (Best): {best_model_path}")
print(f"✓ PyTorch (Last): {last_model_path}")
print()

# 1️⃣1️⃣ Display Training Results
print("=" * 70)
print("TRAINING RESULTS VISUALIZATION")
print("=" * 70 + "\n")

results_dir = "runs/detect/yolo11s_solar_panel"

# Display training curves if they exist
results_img = os.path.join(results_dir, "results.png")
confusion_matrix = os.path.join(results_dir, "confusion_matrix.png")

if os.path.exists(results_img):
    print("📊 Training curves:")
    display(Image(filename=results_img))

if os.path.exists(confusion_matrix):
    print("\n📊 Confusion matrix:")
    display(Image(filename=confusion_matrix))

print()

# 1️⃣2️⃣ Usage Instructions
print("=" * 70)
print("HOW TO USE YOUR TRAINED MODEL")
print("=" * 70 + "\n")

print("🔹 Load and use your model:")
print("```python")
print("from ultralytics import YOLO")
print()
print("# Load the trained model")
print(f"model = YOLO('{best_model_path}')")
print()
print("# Run predictions")
print("results = model.predict('image.jpg', conf=0.25)")
print()
print("# Process results")
print("for result in results:")
print("    boxes = result.boxes  # Bounding boxes")
print("    for box in boxes:")
print("        cls = int(box.cls[0])  # Class ID")
print("        conf = float(box.conf[0])  # Confidence")
print("        xyxy = box.xyxy[0].tolist()  # Box coordinates")
print("        print(f'Class: {cls}, Confidence: {conf:.2f}, Box: {xyxy}')")
print("```\n")

# 1️⃣3️⃣ Final Summary
print("=" * 70)
print("TRAINING SUMMARY")
print("=" * 70 + "\n")

print(f"✅ Model: YOLO11 Small")
print(f"✅ Dataset: {num_classes} classes - {class_names}")
print(f"✅ Epochs: {EPOCHS}")
print(f"✅ Final mAP@0.5: {metrics.box.map50:.4f}")
print(f"✅ Final mAP@0.5:0.95: {metrics.box.map:.4f}")
print(f"\n📁 Output files:")
print(f"   - Best model: {best_model_path}")
print(f"   - Training plots: {results_dir}/")
print(f"   - Predictions: runs/detect/predict/")
print()

# 1️⃣4️⃣ Download Model (Optional)
print("=" * 70)
print("DOWNLOAD MODEL TO LOCAL MACHINE")
print("=" * 70 + "\n")

print("To download the trained model, uncomment and run:")
print()
print("```python")
print("from google.colab import files")
print(f"files.download('{best_model_path}')")
print("```")
print()

# Uncomment to download:
# from google.colab import files
# files.download(best_model_path)

print("=" * 70)
print("🎉 ALL DONE! Your YOLO11 Medium model is trained and ready!")
print("=" * 70)
````
